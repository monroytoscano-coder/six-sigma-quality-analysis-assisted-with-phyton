```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Leer los datos y calcular parámetros (Lo que ya dominas)
df = pd.read_excel("Ejercicio_Control_Calidad.xlsx", header=4)
datos_ph = df["Medición (pH)"]
promedio = datos_ph.mean()
desviacion = datos_ph.std()

# 2. Calcular los límites Six Sigma (Promedio ± 3 Desviaciones)
lcs = promedio + (3 * desviacion)
lci = promedio - (3 * desviacion)

# 3. Configurar el tamaño del reporte visual
plt.figure(figsize=(10, 5))

# 4. Dibujar los elementos del gráfico
plt.plot(datos_ph, marker='o', color='black', label='Nivel de pH')
plt.axhline(promedio, color='green', linestyle='-', label='Línea Central (Media)')
plt.axhline(lcs, color='red', linestyle='--', label='Límite de Control Superior (+3σ)')
plt.axhline(lci, color='red', linestyle='--', label='Límite de Control Inferior (-3σ)')

# 5. Agregar estética corporativa (títulos y etiquetas)
plt.title("Gráfico de Control: Estabilidad del pH en el Reactor")
plt.xlabel("Secuencia de Lotes")
plt.ylabel("Nivel de pH")
plt.legend(loc='upper right')
plt.grid(True, alpha=0.3)

# 6. Ejecutar y mostrar la imagen
# 6. Guardar la imagen en alta resolución y luego mostrarla
# 6. Guardar la imagen en alta resolución y luego mostrarla
plt.savefig("Grafico_Control_pH.png", dpi=300, bbox_inches='tight')
plt.show()
```
